## Google Gemma model exploration summary:
It turns out that Gemma is a greedy model, meaning that we cannot obtain multiple sequences at a time. So to compare the model's legibility as a funcion of model parameters, I decided to tweak the max_length of the response and the temperature. I think that when one increases max_length, there is a higher chance to obtain something more legible. With temperature, decreasing it will allow the model to suggest a response that is more expected. While raising the temperature will make the model opt out for more 'creative' responses that are sometimes not legible at all.

```python
import transformers
from transformers import AutoTokenizer,AutoModelForCausalLM, AutoConfig
from transformers import pipeline

# Prompt batch
prompts = ["My name is Jeff",
          "We live in a society where",
          "I don't understand what",
          "May the force be"]

model_id = "meta-llama/Llama-3.2-1B"
#model_id = "google/gemma-2-2b-it"

pipe_llama = pipeline(
    "text-generation", 
    model=model_id,
    token="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",
    device='cuda')

pipe_llama(prompts, max_length=20, num_return_sequences=5)
```

### Comparing GPT2 with meta-llama/Llama-2-7b-chat-hf

```python
from transformers import AutoTokenizer, AutoModel, utils, AutoModelForCausalLM

from bertviz import model_view
utils.logging.set_verbosity_error()  # Suppress standard warnings

model_name = 'openai-community/gpt2'
input_text = "The bird flew right by"
model = AutoModelForCausalLM.from_pretrained(model_name, output_attentions=True)
tokenizer = AutoTokenizer.from_pretrained(model_name)
inputs = tokenizer.encode(input_text, return_tensors='pt')  # Tokenize input text
outputs = model(inputs)  # Run model
attention = outputs[-1]  # Retrieve attention from model outputs
tokens = tokenizer.convert_ids_to_tokens(inputs[0])  # Convert input ids to token strings
model_view(attention, tokens)  # Display model view
```

```python
utils.logging.set_verbosity_error()  # Suppress standard warnings

model_name = 'distilbert/distilgpt2'
input_text = "The bird flew right by"
model = AutoModelForCausalLM.from_pretrained(model_name, output_attentions=True)
tokenizer = AutoTokenizer.from_pretrained(model_name)
inputs = tokenizer.encode(input_text, return_tensors='pt')  # Tokenize input text
outputs = model(inputs)  # Run model
attention = outputs[-1]  # Retrieve attention from model outputs
tokens = tokenizer.convert_ids_to_tokens(inputs[0])  # Convert input ids to token strings
model_view(attention, tokens)  # Display model view
```

We can see that by using a smaller model such as distilgpt2, the attention mechanisms are reduced. With the larger GPT2 model, there is more connections between different words in the prompt, in comparison to the distilgpt2 model.
